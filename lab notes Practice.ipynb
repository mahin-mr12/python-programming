Palindrome Number

```python
n = 1232
reverse = 0
temp = n
while temp != 0:
    reverse = (reverse * 10) + (temp % 10)
    temp = temp // 10
if n == reverse:
    print("Paindrome")
else:
    print("Not Palindrome")
```

Armstrong Number


```python
num = 153
order = len(str(num))
sum = 0
temp = num
while temp > 0:
    digit = temp % 10
    sum += digit ** order
    temp //= 10
if num == sum:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```
